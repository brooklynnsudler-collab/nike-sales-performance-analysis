## Part 1: Understanding the Data

```python
# Inspecting the dataset
import pandas as pd
import numpy as np

#load the Nike dataset
df = pd.read_csv("../data/raw/Nike_Sales_Uncleaned.csv")

#display first five records
display(df.head())

#number of rows and columns
print(f"Dataset dimensions: {df.shape}")
print()

#column names and data types
df.info()
print()

#count missing values
print("Missing Values:")
print(df.isnull().sum())
```

```python
# Investigating data quality

#numerical summaries
df.describe()

#duplicate records
print(f"Duplicate rows: {df.duplicated().sum()}")
print(f"Duplicate Order IDs: {df["Order_ID"].duplicated().sum()}")

#unusual numerical values
print(f"Zero revenue: {(df["Revenue"] == 0).sum()}")
print(f"Negative revenue: {(df["Revenue"] < 0).sum()}")
print(f"Negative units sold: {(df["Units_Sold"] < 0).sum()}")
print(f"Negative profit: {(df["Profit"] < 0).sum()}")
print(f"Discounts above 100%: {(df["Discount_Applied"] > 1).sum()}")

#unique categories
print("\nProduct Lines:")
print(df["Product_Line"].value_counts())

print("\nSales Channels:")
print(df["Sales_Channel"].value_counts())

print("\nRegions:")
print(df["Region"].value_counts())
print()

#date formats
print(df["Order_Date"].dropna().head(15))
```

```python
# Investigating date formats

#examine all unique formats
print("Unique date values:")
print(df["Order_Date"].dropna().unique()[:30])

#count how many records use each format
date_counts = df["Order_Date"].dropna().value_counts()

print("\nMost common date values:")
print(date_counts.head(20))
print()

#counting dates by their formatting pattern
dates = df["Order_Date"].dropna().astype(str)

print("YYYY-MM-DD:",
      dates.str.match(r"^\d{4}-\d{2}-\d{2}$").sum())

print("YYYY/MM/DD:",
      dates.str.match(r"^\d{4}/\d{2}/\d{2}$").sum())

print("DD-MM-YYYY:",
      dates.str.match(r"^\d{2}-\d{2}-\d{4}$").sum())

print("DD/MM/YYYY:",
      dates.str.match(r"^\d{2}/\d{2}/\d{4}$").sum())
```

```python
# Investigating duplicate Order IDs

#find Order IDs that apppear more than once
duplicate_orders = df[
    df["Order_ID"].duplicated(keep=False)
].sort_values("Order_ID")

display(duplicate_orders.head(15))

#count records associated with each Order ID
order_counts = df["Order_ID"].value_counts()

print("Most frequently repeated Order IDs:")
print(order_counts.head(15))

#inspecting all records belonging to OrderID 2000
print()
display(
    df[df["Order_ID"] == 2000]
    [["Order_ID", "Product_Name", "Order_Date",
      "Sales_Channel", "Region", "Revenue"]]
)

#investigating number of unique OrderIDs
print(f"\nUnique Order IDs: {df["Order_ID"].nunique()}")
print(f"Total records: {len(df)}")
```

```python
# Do repeated Order IDs consistently have conflicting information?

#counting unique dates and channels for each Order ID
order_check = df.groupby("Order_ID").agg(
    record_count=("Order_ID", "size"),
    unique_dates=("Order_Date", "nunique"),
    unique_channels=("Sales_Channel", "nunique"),
    unique_regions=("Region", "nunique")
)

#show repeated Order IDs with conflicting informaton
conflicting_orders = order_check[
    (order_check["record_count"] > 1) &
    (
        (order_check["unique_dates"] > 1) |
        (order_check["unique_channels"] > 1) |
        (order_check["unique_regions"] > 1)
    )
]
print(conflicting_orders)
print(f"\nNumber of conflicting Order IDs: {len(conflicting_orders)}")
```

## Part 2: Data Cleaning

```python
#creating working copy of original dataset
df_clean = df.copy()

#verify that both datasets contain the same records
print(f"Original dataset: {df.shape}")
print(f"Working dataset: {df_clean.shape}")

#check they are separate objects
print(f"Separate copies: {df is not df_clean}")
```

```python
# Cleaning categorical data
print(df_clean["Region"].value_counts())  #inspect current values
print()

region_mapping = {  #create mapping dictionary
    "bengaluru": "Bengaluru",
    "Bangalore": "Bengaluru",
    "hyderbad": "Hyderabad",
    "Hyd": "Hyderabad"
}

df_clean["Region"] = df_clean["Region"].replace(region_mapping)  #apply replacements

print(df_clean["Region"].value_counts())  #verify results
```

```python
# Standardizing dates

df_clean["Order_Date_Clean"] = pd.NaT  #create new date column

#store dates as strings and remove surrounding spaces
dates = df_clean["Order_Date"].astype("string").str.strip()

#identify each date format
mask1 = dates.str.match(r"^\d{4}-\d{2}-\d{2}$", na=False)

mask2 = dates.str.match(r"^\d{4}/\d{2}/\d{2}$", na=False)

mask3 = dates.str.match(r"^\d{2}-\d{2}-\d{4}$", na=False)

#converting the dates
#format 1: YYYY-MM-DD
df_clean.loc[mask1, "Order_Date_Clean"] = pd.to_datetime(
    dates[mask1],
    format="%Y-%m-%d",
    errors="coerce"
)

#format 2: YYYY/MM/DD
df_clean.loc[mask2, "Order_Date_Clean"] = pd.to_datetime(
    dates[mask2],
    format="%Y/%m/%d",
    errors="coerce"
)

#format 3: DD-MM-YYYY
df_clean.loc[mask3, "Order_Date_Clean"] = pd.to_datetime(
    dates[mask3],
    format="%d-%m-%Y",
    errors="coerce"
)

#verify results
#display original and cleaned dates
display(
    df_clean[["Order_Date", "Order_Date_Clean"]].head(20)
)

#check new data type
print(f"\nNew data type: {df_clean["Order_Date_Clean"].dtype}")

#count converted dates
print(f"Successfully converted: {df_clean["Order_Date_Clean"].notna().sum()}")

#count missing or unparsed dates
print(f"Missing or unparsed: {df_clean["Order_Date_Clean"].isna().sum()}")
```

```python
# Investigating missing values

#calculate missing values
missing_summary = pd.DataFrame({
    "Missing_Count": df_clean.isnull().sum(),
    "Missing_Percentage": (
        df_clean.isnull().mean() * 100
    ).round(2)
})
display(missing_summary)

#examine records missing important sales information
print()
display(
    df_clean[
        df_clean["Units_Sold"].isna() &
        df_clean["MRP"].isna()
        ].head(10)
)

# Missing value patterns
#missing value combinations
sales_columns = [
    "Units_Sold",
    "MRP",
    "Discount_Applied"
]

missing_patterns = (
    df_clean[sales_columns]
    .isna()
    .value_counts()
    .reset_index(name="Record_Count")
)
print()
display(missing_patterns)
```

```python
#compare missing sales information with revenue values
df_clean["Missing_Sales_Info"] = (
    df_clean[
        ["Units_Sold", "MRP", "Discount_Applied"]
        ].isna().any(axis=1)
)

revenue_comparison = df_clean.groupby(
    "Missing_Sales_Info"
).agg(
    Total_Records=("Revenue", "size"),
    Zero_Revenue=("Revenue", lambda x: (x == 0).sum()),
    Nonzero_Revenue=("Revenue", lambda x: (x != 0).sum())
)

display(revenue_comparison)
```

```python
# Find records with complete sales information but zero revenue
complete_zero_revenue = df_clean[
    (df_clean["Missing_Sales_Info"] == False) &
    (df_clean["Revenue"] == 0)
]

display(
    complete_zero_revenue[
        [
            "Order_ID",
            "Product_Name",
            "Units_Sold",
            "MRP",
            "Discount_Applied",
            "Revenue",
            "Profit"
        ]
    ].head(15)
)

# Calculating estimated revenue
df_clean["Estimated_Revenue"] = (
    df_clean["Units_Sold"] *
    df_clean["MRP"] *
    (1 - df_clean["Discount_Applied"])
)

#comparing actual and estimated revenue
df_clean["Revenue_Difference"] = (
    df_clean["Revenue"] -
    df_clean["Estimated_Revenue"]
)

display(
    df_clean[
        [
            "Units_Sold",
            "MRP",
            "Discount_Applied",
            "Revenue",
            "Estimated_Revenue",
            "Revenue_Difference"
        ]
    ].head(15)
)
```

```python

# Investigate units sold among complete records that have zero revenue
print("Units Sold distribution:")
print(
    complete_zero_revenue["Units_Sold"].value_counts(
        dropna=False
    )
)

print("\nProfit summary:")
print(
    complete_zero_revenue["Profit"].describe()
)

print("\nRecords with zero units and nonzero profit:")
zero_units_profit = complete_zero_revenue[
    (complete_zero_revenue["Units_Sold"] == 0) &
    (complete_zero_revenue["Profit"] != 0)
]

display(
    zero_units_profit[
        ["Product_Name", "Units_Sold", "MRP",
         "Discount_Applied", "Revenue", "Profit"]
    ]
)

print("Count:", len(zero_units_profit))
```

```python
# Comparing Revenue and Profit 

profit_check = df_clean.groupby(
    df_clean["Revenue"] == 0
).agg(
    Record_Count=("Profit", "size"),
    Average_Revenue=("Revenue", "mean"),
    Average_Profit=("Profit", "mean"),
    Negative_Profit=("Profit", lambda x: (x < 0).sum()),
    Positive_Profit=("Profit", lambda x: (x > 0).sum())
)

display(profit_check)

# Inspecting records with zero revenue but positive units sold

zero_revenue_positive_units = df_clean[
    (df_clean["Revenue"] == 0) &
    (df_clean["Units_Sold"] > 0)
]

display(
    zero_revenue_positive_units[
        [
            "Units_Sold",
            "MRP",
            "Discount_Applied",
            "Revenue",
            "Profit"
        ]
    ].head(15)
)

print(
    "Records:",
    len(zero_revenue_positive_units)
)


# Finding records with positive units and zero revenue
positive_units_zero_revenue = df_clean[
    (df_clean["Units_Sold"] > 0) &
    (df_clean["Revenue"] == 0)
]

#count records with missing MRP
missing_mrp = positive_units_zero_revenue["MRP"].isna().sum()

#count records with missing discounts
missing_discount = positive_units_zero_revenue["Discount_Applied"].isna().sum()

#count records with all sales inputs available
complete_inputs = positive_units_zero_revenue[
    ["Units_Sold", "MRP", "Discount_Applied"]
].notna().all(axis=1).sum()

print("Positive units with zero revenue:",
      len(positive_units_zero_revenue))

print("Missing MRP:", missing_mrp)

print("Missing discount:", missing_discount)

print("All sales inputs available:", complete_inputs)


# Selecting records with nonzero revenue and complete sales inputs

valid_revenue_records = df_clean[
    (df_clean["Revenue"] != 0) &
    (df_clean[
        ["Units_Sold", "MRP", "Discount_Applied"]
    ].notna().all(axis=1))
].copy()

#calculate hypothetical estimated revenue
valid_revenue_records["Estimated_Revenue"] = (
    valid_revenue_records["Units_Sold"] *
    valid_revenue_records["MRP"] *
    (1 - valid_revenue_records["Discount_Applied"])
)

#calculate the difference
valid_revenue_records["Revenue_Difference"] = (
    valid_revenue_records["Revenue"] -
    valid_revenue_records["Estimated_Revenue"]
)

#display comparison
display(
    valid_revenue_records[
        [
            "Units_Sold",
            "MRP",
            "Discount_Applied",
            "Revenue",
            "Estimated_Revenue",
            "Revenue_Difference"
        ]
    ].head(15)
)

print("Records evaluated:", len(valid_revenue_records))

print(
    "Average absolute difference:",
    valid_revenue_records["Revenue_Difference"].abs().mean()
)
```

```python
# Calculate absolute revenue differences
valid_revenue_records["Absolute_Difference"] = (
    valid_revenue_records["Revenue_Difference"].abs()
)

#counting records within one cent
matching_records = (
    valid_revenue_records["Absolute_Difference"] <= 0.01
).sum()

print("Records evaluated:",
      len(valid_revenue_records))

print("Records matching within $0.01:",
      matching_records)

print("Maximum absolute difference:",
      valid_revenue_records["Absolute_Difference"].max())


# Investigate numerical anomalies
print()
print("Negative units sold:",
      (df_clean["Units_Sold"] < 0).sum())

print("Zero units sold:",
      (df_clean["Units_Sold"] == 0).sum())

print("Discounts above 100%:",
      (df_clean["Discount_Applied"] > 1).sum())

print("Negative discounts:",
      (df_clean["Discount_Applied"] < 0).sum())

print("Negative MRP:",
      (df_clean["MRP"] < 0).sum())

print("Negative revenue:",
      (df_clean["Revenue"] < 0).sum())

print("Negative profit:",
      (df_clean["Profit"] < 0).sum())

# Investigating discounts above 100%
high_discount = df_clean[
    df_clean["Discount_Applied"] > 1
]
print()
display(
    high_discount[
        [
            "Order_ID",
            "Product_Name",
            "Units_Sold",
            "MRP",
            "Discount_Applied",
            "Revenue",
            "Profit"
        ]
    ].head(20)
)
print("Total records:", len(high_discount))
print("\nDiscount distribution:")
print(high_discount["Discount_Applied"].describe())


# Flag discounts above 100%
df_clean["Discount_Flag"] = (
    df_clean["Discount_Applied"] > 1
)
#count flagged records
print()
print(df_clean["Discount_Flag"].value_counts())
```

```python
# Create indicators for negative values
negative_units = df_clean["Units_Sold"] < 0

negative_revenue = df_clean["Revenue"] < 0

#compare both conditions
comparison = pd.crosstab(
    negative_units,
    negative_revenue
)
display(comparison)


# Inspect records with negative units and revenue
potential_returns = df_clean[
    (df_clean["Units_Sold"] < 0) &
    (df_clean["Revenue"] < 0)
]
print()
display(
    potential_returns[
        [
            "Product_Name",
            "Units_Sold",
            "MRP",
            "Discount_Applied",
            "Revenue",
            "Profit"
        ]
    ]
)
print("Potential return records:", len(potential_returns))
```

```python
# Reviewing Profit reliability
#compare Revenue and Profit
profit_comparison = df_clean.groupby(
    df_clean["Revenue"] == 0
).agg(
    Record_Count=("Profit", "count"),
    Average_Revenue=("Revenue", "mean"),
    Average_Profit=("Profit", "mean"),
    Negative_Profit=("Profit", lambda x: (x < 0).sum()),
    Positive_Profit=("Profit", lambda x: (x > 0).sum())
)
display(profit_comparison)
```

```python
# Final Data Validation

#check dataset dimensions
print("Final dataset dimensions:", df_clean.shape)

print("Total records:", len(df_clean))
print("Total columns:", len(df_clean.columns))
print()

#check data types
print(df_clean.dtypes)
print()

#check standardized regions
print("Regions:")
print(df_clean["Region"].value_counts())

#check discount flags
print("\nDiscount flags:")
print(df_clean["Discount_Flag"].value_counts())
print()

#missing values 
print("Final missing values report:")
missing_summary = pd.DataFrame({
    "Missing_Count": df_clean.isnull().sum(),
    "Missing_Percentage": (
        df_clean.isnull().mean() * 100
    ).round(2)
})
display(missing_summary)
print()

#check for completely duplicated rows
print(
    "Duplicate rows:",
    df_clean.duplicated().sum()
)

#check date conversion
print(
    "Valid dates:",
    df_clean["Order_Date_Clean"].notna().sum()
)
print(
    "Missing dates:",
    df_clean["Order_Date_Clean"].isna().sum()
)

#check flag columns
print(
    "Unusual discounts:",
    df_clean["Discount_Flag"].sum()
)
```

```python
# Saving the cleaned dataset
df_clean.to_csv(
    "nike_sales_cleaned.csv",
    index=False
)

print("Cleaned dataset successfully saved!")
```

```python
# Create final analysis dataset
df_final = df_clean.drop(
    columns=[
        "Potential_Return",
        "Estimated_Revenue",
        "Revenue_Difference",
        "Order_Date"
    ]
)

#rename cleaned date column
df_final = df_final.rename(
    columns={"Order_Date_Clean": "Order_Date"}
)

#check final dataset
print("Final dimensions:", df_final.shape)
display(df_final.head())
```
